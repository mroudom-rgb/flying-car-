```python
import pandas as pd
from google.colab import files
import os

# Check if the file already exists in the content directory
file_path = "/content/GPU_benchmarks_v7.csv"

if not os.path.exists(file_path):
    print("File not found. Please upload 'GPU_benchmarks_v7.csv'.")
    uploaded = files.upload()
    if 'GPU_benchmarks_v7.csv' not in uploaded:
        raise FileNotFoundError("GPU_benchmarks_v7.csv was not uploaded. Please upload the correct file.")
    # The uploaded file will be in the current working directory, move it to /content if necessary or use the direct path
    # For simplicity, we assume it's directly available after upload in the current working directory or /content
    # If it was uploaded into the current directory, we still use /content/filename as Colab's default for files.upload()

df = pd.read_csv(file_path)

# Show first few rows
print(df.head(10))

cols = list(df.columns)
print("Detected Columns:", cols, "\n")

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        new_node = Node(data)
        if not self.head:
            self.head = new_node
            return
        cur = self.head
        while cur.next:
            cur = cur.next
        cur.next = new_node

    def delete(self, key, col_name):
        cur = self.head

        # Case 1: Head node itself holds the key to be deleted
        if cur and cur.data.get(col_name) == key:
            self.head = cur.next
            return True

        prev = None
        # Search for the key to be deleted, keep track of the previous node
        while cur and cur.data.get(col_name) != key:
            prev = cur
            cur = cur.next

        # If key was not present in linked list
        if not cur:
            return False

        # If key was found, unlink the node from the linked list
        prev.next = cur.next
        return True

    def print_list(self, limit=10):  # ✅ only show first 10
        cur = self.head
        count = 0
        print("=== Singly Linked List (First 10 Items) ===")
        while cur and count < limit:
            print(cur.data)
            cur = cur.next
            count += 1
        if cur:
            print(f"... (showing only first {limit} items)\n")
        else:
            print(f"(Total {count} items printed)\n")


# Create an instance of the SinglyLinkedList
gpu_list = SinglyLinkedList()

# Append rows from the DataFrame to the linked list
for _, row in df.iterrows():
    gpu_list.append(row.to_dict())

# Print first few
gpu_list.print_list()

# Delete a specific GPU entry using the correct column name 'gpuName'
deleted_successfully = gpu_list.delete("GeForce RTX 3090", "gpuName")
print(f"Deletion successful: {deleted_successfully}\n")

# Print again to verify deletion
gpu_list.print_list()
```
